```python
import spacy
from spacy.tokens import DocBin
from spacy.training import Example
from tqdm import tqdm
import json
```

```python
nlp = spacy.load("en_core_web_sm")
if "ner" not in nlp.pipe_names:
    ner = nlp.create_pipe("ner")
    nlp.add_pipe(ner, last=True)
else:
    ner = nlp.get_pipe("ner")

labels = ["GPE", "DISASTER"]
for label in labels:
    ner.add_label(label)

def load_training_data(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        lines = f.readlines()
    training_data = []
    for line in lines:
        entry = json.loads(line)
        text = entry[0]
        entities = entry[1]["entities"]
        training_data.append((text, {"entities": entities}))
    return training_data
```

```python
from spacy.training import offsets_to_biluo_tags

def validate_entities(nlp, text, entities):
    try:

        doc = nlp.make_doc(text)
        biluo_tags = offsets_to_biluo_tags(doc, entities)
        
        if '-' in biluo_tags:
            print(f"Invalid tag '-' found in text: {text}")
            return False
        return True
    except Exception as e:
        print(f"Error aligning entities for text: {text}\nError: {e}")
        return False
    
def create_training_examples(nlp, train_data):
    examples = []
    for text, annotations in tqdm(train_data):
        entities = annotations["entities"]
        if validate_entities(nlp, text, entities):
            doc = nlp.make_doc(text)
            example = Example.from_dict(doc, annotations)
            examples.append(example)
        else:
            print(f"Skipping misaligned data: {text}")
    return examples
```

```python
train_data = load_training_data("output4.jsonl")
```

```python

train_examples = create_training_examples(nlp, train_data[:18700])
```

```python
learning_rate = 0.00001  

other_pipes = [pipe for pipe in nlp.pipe_names if pipe != "ner"]
with nlp.disable_pipes(*other_pipes):
    optimizer = nlp.resume_training()
    nlp.config["training"]["optimizer"]["learn_rate"] = learning_rate
    for iteration in range(40):  
        losses = {}
        batches = spacy.util.minibatch(train_examples, size=spacy.util.compounding(128.0, 256.0, 1.3))
        for batch in batches:
            nlp.update(batch, sgd=optimizer, drop=0.20, losses=losses)
        print(f"Iteration {iteration + 1}, Losses: {losses}")

nlp.to_disk("model_1.2j.2")
# in x.y.z, x=spacy model, y=yth dataset, z=zth version
print("Model fine-tuning complete!")
```
