```python
import matplotlib.pyplot as plt

def plot_epoch_loss_curve(file_path):
    epochs = []
    losses = []

    with open(file_path, "r") as file:
        for line in file:
            if "Iteration" in line and "Losses" in line:
                parts = line.split(", Losses: {'ner': ")
                epoch = int(parts[0].replace("Iteration ", "").strip())
                loss = float(parts[1].replace("}", "").strip())
                epochs.append(epoch)
                losses.append(loss)

    plt.figure(figsize=(7, 5))  
    plt.plot(epochs, losses, marker='o', color='dodgerblue', linestyle='-', linewidth=2, markersize=5, label="Loss")
    plt.xlabel("Epoch", fontsize=14, fontweight='bold', color='navy')
    plt.ylabel("Loss", fontsize=14, fontweight='bold', color='navy')
    plt.grid(True, linestyle="--", alpha=0.7, color='gray', linewidth=0.5)
    plt.xticks(fontsize=12, color='dimgray')
    plt.yticks(fontsize=12, color='dimgray')
    plt.legend(fontsize=12, loc="upper right", frameon=False)
    plt.tight_layout()
    plt.show()

file_path = "loss.txt" 
plot_epoch_loss_curve(file_path)
```

```python
import matplotlib.pyplot as plt

def plot_epoch_loss_curve(file_path):
    epochs = []
    losses = []

    with open(file_path, "r") as file:
        for line in file:
            if "Iteration" in line and "Losses" in line:
                parts = line.split(", Losses: {'ner': ")
                epoch = int(parts[0].replace("Iteration ", "").strip())
                loss = float(parts[1].replace("}", "").strip())
                epochs.append(epoch)
                losses.append(loss)

    plt.figure(figsize=(6, 4))
    plt.plot(epochs, losses, marker='o', label="Loss (NER)")
    
    plt.yscale("log")
    
    plt.title("Epoch vs Loss Curve", fontsize=16)
    plt.xlabel("Epoch", fontsize=14)
    plt.ylabel("Loss (Log Scale)", fontsize=14)
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.legend(fontsize=12)
    plt.tight_layout()

    plt.show()

file_path = "loss.txt"  
plot_epoch_loss_curve(file_path)
```
